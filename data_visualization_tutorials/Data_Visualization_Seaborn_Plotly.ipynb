## **Seaborn**

"Seaborn is a Python data visualization library based on matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics." Source: [Seaborn](https://seaborn.pydata.org)

```python
# Import necessary libraries
import seaborn as sns
from sklearn.datasets import load_iris # The dataset that we will use
```

Load in the Iris dataset for demonstrations

```python
iris = load_iris(as_frame=True).frame
iris['target'] = load_iris().target_names[iris['target']] # This line changes the target to the name of the flowers
iris.head()
```

### **Regression Plot**

A Regression Plot is a scatter plot with a line of best fit drawn. It also has a margin of errors around the line of best fit.

To create a Regression Plot, we use the function `sns.regplot()`. The following parameters are required:
- `data`: The dataframe containing the information to be plotted
- `x`: The column name in the dataframe for the independent variable
- `y`: The column name in the dataframe for the dependent variable

```python
sns.regplot(data=iris, x='sepal length (cm)', y='sepal width (cm)');
```

### **Histogram**
Histograms can be created using `sns.histplot()`. It only needs one variable.

```python
sns.histplot(iris['sepal length (cm)']);
```

A Kernel Density Estimation can be added to the histogram by using the parameter `kde=True`.

```python
sns.histplot(iris['sepal length (cm)'], kde=True);
```

### **Pairplots**

For quick visualization of the entire dataset, `sns.pairplot()` can be used.

In the pairplots, every numerical column is plotted against every other numerical column.
- Along the diagonal, the column is plotted against itself, so histograms are created
- Everywhere else, the scatter plots are created

```python
sns.pairplot(iris);
```

The pairplots can also be separated by the categorical column using `hue = `. The name of the column needs to be given.

By doing so, the histogram is replaced with a multi-layered density plot.

```python
sns.pairplot(iris, hue='target');
```

### **Heatmap**

A heatmap can be created using all the numerical columns to see the correlation coefficient between every pair.

```python
# Get the correlations of all the numerical columns only
iris.corr(numeric_only=True)
```

```python
# Using the heatmap to visually represent the correlations
# annot adds the actual values of the correlations
sns.heatmap(iris.corr(numeric_only=True), annot=True);
```

## **Plotly**

Plotly is the home of beautiful, interactive data visualization. Connect to data, build with AI agents, and share your analytics everywhere. - [Plotly](https://plotly.com/)

The advantage of Plotly is that the visualizations are interactive.
- Data can be displayed by hovering cursors over the elements
- Individual elements in the plot can be selected or de-selected

There are several icons that are useful in Plotly
- The camera icon can be used to download the visualization as a png file
- The magnify glass icon can be used to zoom in or out of the visualization but selecting an area
- The arrows icon can be to move around on the visualization
- The dotted square icon is box select
- The circle icon is the lasso select
- The plus and minus icons are zoom in and zoom out
- The diagonal arrows icon is for autoscaling
- The house icon is to reset the visualization

```python
# Import needed libraries
import pandas as pd
import plotly.offline as py
import plotly.graph_objs as go

py.init_notebook_mode(connected=True)
```

Reading the crime dataset for visualization purposes

```python
crime = pd.read_csv('crime-new.csv', parse_dates=['occurrencedate'])
crime.head()
```

### **Bar Plot**

An interactive bar plot can be created using the MCI types and counts

```python
# Getting the data for the bar plot
crime_types = crime['MCI'].value_counts()

crime_types
```

```python
# Creating the data object, used for plotting
data = go.Bar(x=crime_types.index,
             y = crime_types.values)

# Displaying the figure
go.Figure(data)
```

### **Layout**

The title and axis labels can be added by using `go.Layout()`.
- title parameter asks for a string
- xaxis and yaxis parameters ask for a dictionary

```python
# Creating a layout
layout = go.Layout(title = 'MCI Counts',
                  xaxis = {'title':'Type'},
                  yaxis = {'title':'Count'})

# Adding the data and layout to the figure
go.Figure(data,layout)
```

### **Box Plot**

A box plot can be created using `go.Box()`.

A parameter that is not used in the example is `boxpoints`. This parameter can be set to `boxpoints = 'all'` so that all the points are also plotted in the visualization.

```python
# Creating the data by identifying the x and y variables
data = go.Box(x=iris['target'],
             y= iris['sepal length (cm)'])

# Displaying the figure
go.Figure(data)
```

### **Multiple-Traces**

Plotly graphs can also be laid on top of each other.

```python
# Cleaning the MCI data
crime = crime[crime['occurrenceyear']>=2014]
crime.head()
```

```python
# Creating the datasets for robbery and auto theft
robbery_date = (crime[crime['MCI']=='Robbery']
               .groupby(['occurrenceyear','occurrencemonth'], as_index = False)
               .agg({'occurrencedate':'min', 'MCI':'count'})
               .rename(columns={'MCI':'count'})
               .sort_values('occurrencedate'))

auto_date = (crime[crime['MCI']=='Auto Theft']
               .groupby(['occurrenceyear','occurrencemonth'], as_index = False)
               .agg({'occurrencedate':'min', 'MCI':'count'})
               .rename(columns={'MCI':'count'})
               .sort_values('occurrencedate'))
```

```python
# Creating the robbery data
robbery = go.Scatter(x=robbery_date['occurrencedate'],
                    y=robbery_date['count'],
                    name='Robbery')

# Creating the auto theft data
auto = go.Scatter(x=auto_date['occurrencedate'],
                  y=auto_date['count'],
                 name = 'Auto Theft')

# Creating the layout of the figure
layout = go.Layout(title='MCI Trends',
                  xaxis = {'title': 'Occurrence Year'},
                  yaxis = {'title': 'Count'})

# Displaying the plots together by putting the data in a list
go.Figure([robbery, auto], layout)
```

With multiple graphs layered together, a legend is displayed on the right hand side. You can click on the items in the legend to hide or show the plot.

## **Example 1**

Using a for loop, make a line chart with all the MCI crimes.

```python
# First, create a list of MCI
mcis = crime['MCI'].unique()
mcis
```

```python
# Create an empty list of all_data
all_data = []
```

```python
# Follow the same structure as above and use for loop to create the data from each
for mci in mcis:
    crime_date = (crime[crime['MCI']==mci]
                  .groupby(['occurrenceyear', 'occurrencemonth'], as_index = False)
                  .agg({'occurrencedate': 'min', 'MCI':'count'})
                  .rename(columns={'MCI':'count'})
                  .sort_values('occurrencedate'))

    data = go.Scatter(x = crime_date['occurrencedate'],
                     y = crime_date['count'],
                     name = mci)

    all_data.append(data)
```

```python
# Creating the layout
layout = go.Layout(title = 'MCI Trends',
                  xaxis = {'title':'Occurrence Year'},
                  yaxis = {'title': 'Count'})
```

```python
# Display the figure
go.Figure(all_data, layout)
```

## **Example 2**
Using a new dataset (school earnings dataset), compare the men and women earnings. All include the wage gap.

```python
# Getting the earnings dataset from an internet source
earnings = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/school_earnings.csv")
earnings.head()
```

```python
# Creating the bar plot data
data = go.Bar(
    x=earnings['School'],
    y=earnings['Gap']
)

# Creating the layout for the visualization
layout = go.Layout(
    title="Average Earnings for Graduates",
    xaxis=dict(title='School'),
    yaxis=dict(title='Salary (in thousands)'))


go.Figure(data=[data],layout=layout)
```

The data can be separated by the gender and gap, and displayed together by the schools.

```python
# Creating the women data
trace_women = go.Bar(
    x=earnings.School,
    y=earnings.Women,
    name='Women', # Giving the name of the label
    marker=dict(color='#ffcdd2')) # Defining the color of the plot

# Creating the men data
trace_men = go.Bar(
    x=earnings.School,
    y=earnings.Men,
    name='Men',
    marker=dict(color='#A2D5F2'))

# Creating the gap between the genders
trace_gap = go.Bar(
    x=earnings.School,
    y=earnings.Gap,
    name='Gap',
    marker=dict(color='#59606D'))

# Creating a list of the data
data = [trace_women, trace_men, trace_gap]

# Defining the layout of the plots
layout = go.Layout(
    title="Average Earnings for Graduates",
    xaxis=dict(title='School'),
    yaxis=dict(title='Salary (in thousands)'))

# Displaying the figure
go.Figure(data=data, layout=layout)
```

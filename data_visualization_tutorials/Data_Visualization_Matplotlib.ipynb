## **Basics**



```python
# Import the library
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

First, we need to create a simple dataset to use for plotting

```python
# Data for plotting
np.random.seed(27)        # reproducible

length = 20
noise_variance = 0.1      # Variance of the gaussian noise

noise = np.random.randn(length) * noise_variance

x = np.random.uniform(0, 1,length)
y = x*2 + noise

print(f"x is {x}")
print("----------")
print(f"y is {y}")
```

### **Basic Plot**
To create a plot, we just need to use `plt.plot(x,y)`.

Here `x` is the independent/explanatory variable and `y` is the dependent/response variable.

By default, `plt.plot()` creates a line plot.

```python
# Simple plot

plt.plot(x,y);
# We add ; to omit some system outputs that are not necessary for data analysis
```

### **Markers**

To update the marker, we add a string parameter in pyplot, `plt.plot(x,y,'o')`.

### **Adding Layers**

Sometimes, it is more informative to lay one graph on top of another for comparison. This can be done by adding layers.

```python
# Plot 1
plt.plot(x,y,'ro');
```

```python
# Plot 2
plt.plot([0,1], [0, 2]);
```

```python
# Putting them together
plt.plot(x,y,'ro')
plt.plot([0,1], [0, 2]);
```

As it can be seen, layering the graphs can be done by putting the multiple `plt.plot` together in the same code block. Also, `;` is only needed at the last graph to omit the system output messages.

### **Transparency**

When putting graphs together, sometimes the colours make the resulting graph difficult to read. Consider the following graph.

```python
# Bigger dataset
np.random.seed(37)

length = 1000
noise_variance = 0.2

noise = np.random.randn(length) * noise_variance

x2 = np.random.uniform(0, 1,length)
y2 = x2*2 + noise
```

```python
plt.plot(x2, y2, 'ro')
plt.plot([0, 1], [0, 2]);
```

In the above graph, the line plot (or line of fit) is difficult to see.

To fix this, we can change the transparency of the scatter plot. To do this, we use the `alpha` parameter.

The `alpha` parameter takes a value between 0 and 1. Think of this number as percentage of saturation.

```python
plt.plot(x2, y2, 'ro', alpha = 0.2)
plt.plot([0, 1], [0, 2]);
```

### **Size/Width**

We can also change the size and width of the markers. <b>Note</b>: width is only used for line graphs.

- To change the size, we use the `markersize` parameter.
- To change the width, we use the `linewidth` parameter.

Both parameters takes a positive value.

```python
plt.plot(x,y,'x', markersize = 10);
```

```python
plt.plot([0,1], [0,2], linewidth=10);
```

### **Adding a Title, Labels, and a Legend**

To make the graphs more informative, we can add titles and labels to the graphs.

To add titles, we will use `plt.title()` while the labels are adding using `plt.xlabel()` and `plt.ylabel()`

```python
plt.plot(x2, y2, 'x')
plt.title('First Plot')
plt.xlabel('Independent')
plt.ylabel('y'); # Adding ; to the last line to omit system outputs
```

For multiple layers, we can add a legend to the graph.

To add the legend, we need to:
1. Add the parameter `label` to each `plt.plot()`
2. Use the code `plt.legend()`

The position of `plt.legend()` can be changed to one of the following using the `loc` parameter:
- upper left
- upper center
- upper right
- center left
- center
- center right
- lower left
- lower center
- lower right
By default, the legend is at *upper left*.

```python
plt.legend?
```

```python
plt.plot(x2, y2, 'rx', alpha = 0.3, label = "plot1")
plt.plot([0,1],[0,2], label = 'plot2')
plt.title('First Plot')
plt.xlabel('Independent')
plt.ylabel('y')
plt.legend(loc="lower right");
```

### **Adding Gridlines**

To make the points easier to read, a gridline can be added to the graph.

`plt.grid()` also takes the `alpha` parameter to change the transparency.

```python
plt.plot(x2, y2, 'rx', alpha = 0.3, label = "plot1")
plt.plot([0,1],[0,2], label = 'plot2')
plt.title('First Plot')
plt.xlabel('Independent')
plt.ylabel('y')
plt.legend(loc="lower right")
plt.grid(alpha=0.5);
```

The order of the lines of code can be changed without changing the output. However, the order demonstrated above is the recommended order.

```python
# Changing the order of the code
plt.title('First Plot')
plt.ylabel('y')
plt.plot(x2, y2, 'rx', alpha = 0.3, label = "plot1")
plt.xlabel('Independent')
plt.plot([0,1],[0,2], label = 'plot2')
plt.legend(loc="lower right")
plt.grid(alpha=0.5);
```

## Subplots

Sometimes layering graphs on top of each other is not the best approach. When comparing different data, it can be better to show the plots side by side. This is where subplots come in.

To create subplots, there are two important variables to note
- fig (the overall figure of the entire collection of graphs)
- ax (used to access the individual graph in the collection)

We can use the code `fig, ax = plt.subplots(rows, columns)` where `rows` and `columns` are integers.

```python
# Empty structure of subplots
fig, ax = plt.subplots(1,2);
```

### **Figure Size**
The figure size can be change to accommodate the display of the graphs. It asks for a tuple `figsize = (width, height)` where both `width` and `height` are integers.

```python
# Empty structure of subplots
fig, ax = plt.subplots(1,2, figsize=(15,5));
```

### **Working with Individual Plots**

The plots are ordered in the `ax` object as a two-dimensional array, `[row, column]`.

We replace `plt` with `ax[row, column]`.

**Remember**: Python is a *0-indexed* language meaning we count starting at 0.

```python
# Simple plot to use
plt.plot(x,y, 'x');
```

```python
fig, ax = plt.subplots(2,3, figsize=(10, 5))
ax[1,2].plot(x,y,'x');
```

And each graph possess all the attributes as shown before (in section 1).

However, we need to use `.set_title` instead of `.title`. (Similarly for the other labels)

```python
# Adding other elements
fig, ax = plt.subplots(2,2, figsize = (10, 10))
ax[0,0].plot(x,y, 'x')
ax[0,0].set_title('First Plot')
ax[0,0].set_xlabel('x')
ax[0,0].set_ylabel('y')

ax[1,1].plot([0,1],[0,2])
ax[1,1].set_title('Fourth Plot')
ax[1,1].set_xlabel('Independent')
ax[1,1].set_ylabel('Dependent')
```

## **Statistical Plots**

Matplotlib also consists of many commonly used statistical plots. We will go through some of them in this notebook but others can be found [here](https://matplotlib.org/stable/plot_types/stats/index.html)

As a demonstration, we will use the Iris dataset from the sklearn library.

The histogram can be accessed through `plt.hist(data)`.

By default, it organizes the data into 10 parts (or bins)

```python
plt.hist(iris['sepal length (cm)']);
```

The number of bins can be changed by using the `bins` parameter.

```python
plt.hist(iris['sepal length (cm)'], bins=8);
```

The colors of the bars can be changed using the `color` parameter. However, this changes all the bars, not individual bars.

```python
plt.hist(iris['sepal length (cm)'], bins = 8, color = 'red');
```

We can also create horizontal histograms by using the `orientation` parameter.

```python
plt.hist(iris['sepal length (cm)'], bins = 8, orientation='horizontal');
```

### **Boxplots**

Boxplots are created using the code `plt.boxplot(data)`.

```python
plt.boxplot(iris['sepal width (cm)']);
```

### **Pie Charts**

A pie chart can be created using `plt.pie(data)`

```python
plt.pie([1,2,3,4]);
```

**Note**: Pie charts are only useful for categorical data, not numerical/quantative data.

## **More Advanced Example**

For this example, we will use the knowledge we learned in this notebook to create a more meaningful set of graphs.

```python
# Data
iris.head()
```

### **Task**

Make a subplot with the following two graphs (two graphs in one row)
1. Plot the `sepal length` vs `sepal width` of each type of flowers (all in one graph)
2. Plot the `petal length` vs `petal width` of each type of flowers (all in one graph)

```python
# Determining the type of flowers
iris['target'].value_counts()
```

### **Planning**

This task requires two graphs in total, but each graph has three layers (one for each type of flower).

One approach would be to create the graphs of each type of flower separately then layer them together. But this would require 6 graphs (3 for the sepal comparison and 3 for the petal comparison).

Instead, we will be using for loops to make it easier.

```python
# Step 1, create a list of the flower names
names = iris['target'].unique()
names
```

```python
# Step 2, create a list of markers (to differentiate between the flowers)
markers = ['o', 'x', 'v'] # Can be any three markers as discussed previously
```

```python
# Step 3, creating a function to make code simpler
def get_flower_data(flower, col_name, df=iris): #Asking the user to input the flower name and the interested column
    return df[df['target']==flower][f'{col_name} (cm)'] #Uses dataframe parsing to get the desired data
```

```python
# Step 4, iterating through the list of names and markers to create the plot

# Creating the subplot structure
fig, ax = plt.subplots(1, 2, figsize=(10,5))

# Creating the graph for sepal_length vs sepal_width
for name, marker in zip(names, markers): # zip pairs the elements in the lists together following the same order
    sepal_length = get_flower_data(name, 'sepal length') # Get the data for sepal_length
    sepal_width = get_flower_data(name, 'sepal width') # Get the data for sepal_width

    # Plotting the data using the marker and labelling with the flower name
    ax[0].plot(sepal_length, sepal_width, marker, label = name)

# Adding labels to the first graph
ax[0].set_xlabel('Sepal Length')
ax[0].set_ylabel('Sepal Width')

# Adding legend to the graph
ax[0].legend(loc='lower right')

# Repeat the same for the second graph
for name, marker in zip(names, markers):
    petal_length = get_flower_data(name, 'petal length')
    petal_width = get_flower_data(name, 'petal width')

    ax[1].plot(petal_length, petal_width, marker, label = name)

ax[1].set_xlabel('Petal Length')
ax[1].set_ylabel('Petal Width')
ax[1].legend(loc='lower right')

# Creating an overall title for the subplots with the suptitle method
fig.suptitle('Iris')

# Making the output look nicer with the tight_layout() method
fig.tight_layout();
```

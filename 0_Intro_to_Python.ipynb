## **What is Python?**

In its most basic form, a programming language is a set of human-readable instructions that can be used to direct a computer to perform a specific task. Each programming language, like different spoken languages, has its own set of rules regarding syntax and structure.

Python is a programming language whose syntax has been designed to be simple, user-friendly and accessible.

For example lets compare the code to save infromation from a text file into a variable in Python, and another programming language C++.

**Python:**

```python
with open("data.txt") as f:
    content = f.read()
```

**C++:**

```cpp
#include <fstream>
#include <sstream>
#include <string>

int main() {
    std::ifstream file("example.txt");
    std::stringstream buffer;
    buffer << file.rdbuf();
    std::string contents = buffer.str();
}
```

Don't worry too much about understanding the code above, but you can see that the Python code is simpler, shorter, and easier to understand than the C++ program.

## **How to use Python**

There are a few different ways that you can use Python.

Python instructions are typically contained in text files called scripts, and you can install and use a [variety of text editors](https://www.techradar.com/best/best-text-editors) on your computer to edit these scripts. These scripts can then be run by the [Python interpreter](https://www.python.org/downloads/), a program which also would need to be downloaded and installed on your computer. Python is available for free on a variety of different operating systems including [Windows](https://www.python.org/downloads/windows/), [macOS](https://www.python.org/downloads/macos/), and [Linux/Unix](https://www.python.org/downloads/source/).

While installing and working with Python on your own computer is relatively simple, the process can sometimes feel daunting for beginners. A helpful alternative is to work within a web-based system.

Jupyter Notebooks are interactive documents that allow you to combine executable Python code with descriptive text and in this tutorial series we will use the University's [JupyterHub platform](https://datatools.utoronto.ca/) to work with them. With browser-based notebooks, instead of running Python instructions on your own computer, your code is executed online.

Using online notebooks has several advantages in that it involves little setup, facilitates collaboration and allows for easy portability across devices, but there are drawbacks including the inability to work offline, potential file size and memory limitations, and reduced configuration control.

## **Working with Jupyter Notebooks**

Jupyter Notebooks consist of a collection of cells, which can be thought of as distinct blocks of content. The two main types of cells are:
- **Markdown**: for formatted textual information like titles, bullet points, links, or short explanations
- **Code**: for writing and running Python code

In Jupyter Notebook, there are two distinct modes for working with the notebooks: **Command Mode** and **Edit Mode**.

In **Command Mode**, you can manage the notebook itself and use certain shortcuts to edit the notebook. For example, in **Command Mode** you can rearrange the order of your cells or change cell types. To determine that you are in **Command Mode**, the current cell will have a <font style="color:blue">**blue**</font> &#x1F535; bar to the left.

In **Edit Mode**, you are editing the content of the cells and can write and edit code or text in the individual cells. To determine that you are in **Edit Mode**, the current cell will have a <font style="color:green">**green**</font> &#128994; bar to the left.

To enter **Command Mode** either press Esc or click on the left edge of a cell, outside of the text area.

To enter **Edit Mode** either press Enter or click inside the text area of a cell.

### **Kernels**

#### What is a Kernel?
A **kernel** can be thought of as the brain of your notebook. When using a Jupyter Notebook, a kernel is needed in order to store your variable values, execute your Python code and provide results. The kernel is automatically started when a notebook is opened.

#### When would a kernel die?
Every kernel has its own computational power and limits. When too much information, or data, is stored or ran, the kernel might not be able to handle it. When this happens, the kernel "dies", or terminates. Another reason for a kernel to die is when a code is stuck in a (seamingly) infinite loop. Kernels also die, if left unused over a certain period of time.

#### What happens when a kernel dies?
When a kernel dies, the kernel stops running and all variable values stored in the kernel are lost. To continue working, you must re-run the cells.

### **Notebook Linearity**

In a notebook, cells can be run in any order, but it is a good practice to **run cells in sequential order**.

This becomes particuarily important, when you take a break from working on a notebook. Often when you return to your notebook after a bit of time, your kernel will have died. In this scenario, it is recommended that you use the <i class="fa fa-fast-forward"></i> button to start a new kernel and re-run the entire notebook. This functionality is also helpful when working on notebook sent to you by a colleague.

As you work on your notebook, pay attention to the order of your cells. If a newly-created cell contains code that is needed earlier in your notebook, then use the <i class='fa fa-arrow-up'></i> button to move the cell up. Similarly, if a cell should be further down the notebook, use the <i class='fa fa-arrow-down'></i> button.


### **Creating Checkpoints and Saving Your Work**

JupyterHub doesn’t have a traditional undo function when working with notebooks, so be careful when making significant changes to your cells. However, it does allow you to create checkpoints by pressing either **Ctrl-S** or selecting **File→Save and Checkpoint**. Once a checkpoint is created, you can revert your notebook to its last save point, by using **File→Revert to Checkpoint**. Keep in mind that you can only revert to your most recent checkpoint. Another way to save your progress is to download the notebook to your own computer as you work. You can do this by selecting **File→Download as→Notebook (.ipynb)**.

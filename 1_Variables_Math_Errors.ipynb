## **Variables and Data Types**

Central to most work with programming languages is data. For example, you might use Python to analyze writing from a historical text, perform calculations on your laboratory experiment results, or generate statistics on government census information.

To work with this data in a programming language it needs to be stored in a container called a variable. In Python each variable has a data type, which describes the kind of information that is contained within it.

We will learn about a few other data types throughout these tutorials but the three main types in Python are **integers, floats, and strings**:

- **Integers** are whole numbers with no decimal points. Examples: 23, 2, -15
- **Floats** are numbers that can contain decimals. Examples: 12.5, 0.5, -5.123
- **Strings** are collections of textual characters. Examples: "University of Toronto 1995", "science", "Hello World"

To use a variable it needs both a name and a value, and when you give a variable a name and value you are defining a variable. Regardless of the data type, the format of how you perform this definition is the same: the variable name, an equal sign and then the value of the variable. Let's look at some examples:


```python
# examples of variables

age = 12
temperature = 12.2
_fullName = "Bob Smith"
```

We will go through working with strings in more detail in a later tutorial but note when creating a string variable, the string is contained within quotation marks. Those strings can be contained in either single or double quotes, but not both in the same string definition. When creating numeric variables, quotation marks are not used.

If you would like your string to contain a quotation mark, one simple way to handle this situation is to use the opposite type of quotation mark to contain the string, as seen below:

```python
firstString = "My String"
secondString = 'Another String'
quotation = 'He then said, "It is a beautiful day outside"'
```

There are a few rules regarding naming variables in Python:

-	Variable names must begin with a letter or an underscore but cannot begin with a number or other special characters. Numbers can be contained later in the variable name.
-	Python variable names are case sensitive. For example, MyName is a different variable than myname.
-	Variable names cannot contain Python keywords. We will learn Python keywords as we proceed through our tutorials, but Python reserves certain keywords for its own use (Examples: if, else, class, return, global) and these keywords cannot be used as variable names.
-	Spaces are not allowed in variable names.

## **Outputting Multiple Variables at a Time**

We have seen how you can print a single variable at a time, but Python also provides a few different ways to print multiple variables at once.
A simple way to print multiple string variables at once is to use string concatenation by using the + character to join multiple strings together:




```python
city = "Toronto"
province = "Ontario"
location = city + "," + province
print(location)
```

A drawback of this approach is that it can only be used with strings. A more flexible method that allows you to print a collection of variables of different data types are **f-strings**. To use f-strings you precede a string with the letter f, and place variables names inside of { } brackets as seen below:

```python
bowler = "Bob Smith"
score = 300
print(f"Today {bowler} rolled a {score}.")
```

Note in the above example the variable inside the brackets is placed in the string.

F-strings can also be used in variable definitions:



```python
studentName = "Fred Smith"
major = "geography"
university = "University of Toronto"

sentence = f"{studentName} is a {major} major at the {university}."

print(sentence)
```

## **Working with Errors**

An error occurs in Python when you ask the language to do something it cannot. When Python encounters an error, it stops your program and won't run it until you fix it. One of the great things about Python is that its error messages are often quite descriptive and will give you a sense of what might be wrong with your code. Let's look at an example:

```python
temperature = 12
dayOfTheWeek = "Thursday"

forecast = "It will be " + temperature + " degrees " + " on " + dayOfTheWeek

print(forecast)
```

You can see from the error message that Python encountered a **type error** which occurs when you use the wrong data type for an operation. As we learned earlier, string concatenation can only be performed with strings, and temperature is an integer. The error message highlights the line on which the error occurred and provides some background on why the error happened. Let's look at another example:

```python
mathGrade = 85
chemistryGrade = 70

averageGrade = (mathGrade + chemistryGrade + englishGrade) / 3

print(averageGrade)
```

This code resulted in a **name error**, as we tried to use a variable that was not defined. In this case, as the error message indicates, we did not define a variable named englishGrade.

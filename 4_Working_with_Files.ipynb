## **Data Formats**

Python can work with a variety of data formats and before looking at some code, let's go through some of the more common formats you may encounter in your research or coursework:

### **Text Files**

A text file is, as the name suggests, a file that contains exclusively text. The file contains no special formatting that could separate elements of the file into individual components.  It is a common convention that text files have the file extension .txt

File extensions are short 3- or 4-character suffixes preceded by a dot that you may see at the end of a filename.  They can help your operating system understand what type of a file it is, and which program might open it. Example: report.txt

### **CSV**

A CSV file (which stands for comma separated values) is a special type of text file which allows for the storing of data in a table-like format. Fields are typically separated by a comma, and generally one line of text represents one record or row of data. The first line of a CSV file often contains the names of the columns but is not a requirement. CSV files typically have the files extension .csv Example: phoneRecords.csv

Here is a sample of the format of a typical CSV file:

```csv
item,quantity,price
apples,23,1.99
toothpaste,12,3.20
bread,50,4.99
```




### **JSON**

JSON files are also text files, but their structure is organized in a key-value pairs inside objects. With its roots in the JavaScript programming language, JSON stands for JavaScript Object Notation, the data format is now used by many programming languages, especially for configuration files and data exchange.
A simple example of the JSON data format looks almost identical to a Python dictionary:

```json
{
  "name":"Fred",
  "age":22,
  "city":"Toronto"
}
```

Like Python dictionaries, JSON files can store text, numeric values, lists, Boolean values and nested objects as seen in this more complex example:

```json

{
  "students": [
    {"name": "Alex", "grade": 90},
    {"name": "Maria", "grade": 85}
  ]
}
```

JSON files generally have the file extension .json

These three file types are just a small sample of the files that Python can work with. Other formats that you can potentially work with using Python include: [audio files](https://github.com/jiaaro/pydub), [images](https://pillow.readthedocs.io/en/stable/), [XML files](https://realpython.com/python-xml-parser/) and even [Excel spreadsheets](https://openpyxl.readthedocs.io/en/stable/) and [SPSS data files](https://pandas.pydata.org/docs/reference/api/pandas.read_spss.html).  

## **Working With Files in Python**

As you start working more with Python you will find that it's often easier to work with files using helpful libraries, such as [Pandas](https://pandas.pydata.org/), but is important to also understand how to work with files directly with Python's built-in functions.

### **Reading and Writing to Text Files**

Python provides built-in functionality to work with text files as seen in the example below, where we read the entire contents of a text file into a variable:





````python
with open("sample.txt","r") as file:
  contents = file.read()

print(contents)
````

Let's take a closer look at this code line  by line:
- **open("sample.txt","r") as file:** This is a function that tells Python to open the file **sample.txt**. The **"r"** stands for read mode which tells Python that you want to only read the data but not change it. The function creates a file object, and assigns it the variable named file. **with** is what Python calls a context manager, and in this case it is important because it makes sure that your file is closed when you are done working with it.
- **contents = file.read()** The **read()** method tells Python to read every character in the file, and store those characters in a new variable called **contents**.
- **print(contents)** This outputs the variable **contents** to the user.

In addition to reading data from a file, you can also write text to a file, as seen in the example below:

````python
with open("example.txt","w") as file:
  file.write("This is some text for my file. ")
  file.write("Here is a bit more text for the file.")
````

In the above example, the **"w"** indicates that you are writing to the file specified. If the file does not exist in your current directory, the file is created. If the file does exist in the directory, its contents are overwritten.

The **write()** method takes the string inside the parentheses and places it in the file. The first **write()** method in this code places the text at the beginning of the file, and the second **write()** method places its text right after the first written text without creating a new line.

If you would like to add content to a text file without erasing existing content, you can use the append file mode, by using an **"a"** in the open method, as seen below:

````python
with open("example.txt","a") as file:
  file.write("This is new text for the file ")
````

### **Absolute vs. Relative File Paths**

Thus far in our examples we have been working with files that are contained in the same directory as the script we are running, but you may find that you need to work with files contained in other directories and you can do this by using file paths.  There are two forms of file paths in Python: **absolute** and **relative**.

Absolute file paths give the full and exact location of file, starting with the root of the file system:


````python
filePath = "C:/myPrograms/sample.txt" # an example of absolute path in windows

macFilePath = "/myPrograms/sample.txt" # an example of absolute path in macOS / Linux
````

While with Windows systems you would typically use backslashes in directory names, Python allows you to use forward slashes, as seen above. This can be helpful as backslashes are considered special characters in Python.

Relative file paths describe a file's location relative to the current directory you are working in. Our earlier examples were technically examples of a relative file location, but they can also be used with subdirectories:

````python
with open("data/example.txt", "r") as file:
    contents = file.read()
````

You can use **..** notation to move up a directory to a parent directory from the current directory:

````python
with open("../example.txt", "r") as file:
    contents = file.read()
````

Generally, particularly when using Jupyter Notebooks, it is recommended to use relative paths as they more portable across systems and between users.

### **Reading and Writing to CSV Files**

The Python standard library allows for reading of and writing to CSV files. There are a few different ways to read and write CSV files using Python, and we will only focus on a few of the simpler methods. Let's start by reading CSV file information into a list:

````python
import csv

with open("sandwiches.csv","r") as file:
  reader = csv.reader(file)

  header = next(reader)
  print(f"The CSV header is {header}.")

  for row in reader:
    name = row[0]
    price = row[1]
    quantity = row[2]

    print(f"The price of the {name} sandwich is {price}. We have {quantity} available.")
````

Let's go through this script in more detail:
- The **import csv** line allows you to use the csv functions from Python's standard library. Since it is from the standard library, you do not need to use a pip command to install this library.
- The **with open("sandwiches.csv") as file:** line works similar to our text file examples, it opens the sandwiches.csv file in read-only mode.
- The **reader = csv.reader(file)** line creates a reader object in Python, that allows us to work with the formatted CSV file.
- Because our CSV file in this example has a header column with the column names, we want to skip this row when we are printing our row information. With the **header = next(reader)** line you "consume" that first line and store it in the variable header. This moves our file reading cursor to the second line of the file which contains our actual sandwich data.
- Within the **for loop**, we then read each line of the file into the variable row, and access each item of each row by using list indexing.

Particularly if your CSV file has a header row with column names, it can be useful to read your CSV data into a dictionary as seen below:




````python
import csv

with open("sandwiches.csv","r") as file:
  reader = csv.DictReader(file)

  for row in reader:
    print(f"The price of the {row['name']} sandwich is {row['price']}. We have {row['quantity']}")
````

The key difference with the above section of code is that **csv.DictReader()** assumes the first row of your CSV contains headers, and it uses that first row of headers to create a dictionary with those column names as keys for each row.

One important thing to keep in mind when working with CSV files is that data read using these methods is always interpreted as strings. If you want to perform mathematical operations on this data, you must first convert the values to an appropriate numeric type (such as int or float). Fortunately, Python makes this conversion process straightforward:

````python
number1 = int("12")
number2 = int("13")

total = number1 + number2

print(total)
````

In the above example, to convert from a string to integer we just need to precede our string value with the keyword **int**. You can also do this form of conversion with floats as seen in the example below:



````python
number3 = float("12.5")
floatTotal = 100 + number3
print(floatTotal)
````

Here is an example of using this conversion process with our CSV file:

````python
import csv

totalAmount = 0

with open("sandwiches.csv","r") as file:
  reader = csv.DictReader(file)

  for row in reader:
    totalAmount = totalAmount + int(row["quantity"])

print(f"The total number of sandwiches available is {totalAmount}.")

````

The CSV library also allows you to write data to a file in CSV format, again there are a few different ways to do this. Let's start with writing a list to CSV file:

````python
import csv

students = [
        ['name','major','gpa'],
        ['Jim','Math',4.0],
        ['Mary','Chemistry',2.7],
        ['Bob','Physics',3.8]
]

with open("students.csv","w",newline="") as file:
  writer = csv.writer(file)
  writer.writerows(students)
````

In the above example, we have created a list of list items, the first list item represents the header fields of our CSV file, and the subsequent list items are individual student records. Note in our **with open()** code we have included **newline=""**, which helps with some compatibility issues across different operating systems. **csv.writer(file)** creates a csv file object that we can write to, and assigns it the name writer. **writer.writerows(students)** writes all of the rows in our students list to our CSV file.

After running this code block, you should have CSV file containing the student data in your directory.

Similar to reading from a dictionary you can also write a dictionary to CSV file:

````python
import csv

streets = [
    {'name':'Bloor','length':23,'city':'Toronto'},
    {'name':'Queen','length':15,'city':'Toronto'},
    {'name':'Woodlawn','length':10,'city':'Guelph'},
    {'name':'King','length':23,'city':'Hamilton'}
]

with open("streets.csv","w",newline="") as file:
  fields = ['name','length','city']
  writer = csv.DictWriter(file,fieldnames=fields)
  writer.writeheader()
  writer.writerows(streets)
````

In this example we have created a list of dictionary items, and have written this dictionary as rows to our CSV file. Note, with this example we have created a list of the header column values, and used this list with **csv.DictWriter()** to specify what our field names are. With this set, we can use **writeheader()** to write the header row of our CSV file, and **writerows()** to write all of the individual rows of our CSV file.

Note with each of these examples, because we are using the **"w"** flag, we are creating a new CSV file each time. If your file already exists in your directory, its contents will be overwritten. If you would like to add to an existing CSV file you can use the **"a"** flag as seen below:

````python
import csv

newStudents = [
        ['Fred','English',3.3],
        ['Christine','Physics',3.9]
]

with open("students.csv","a",newline="") as file:
  writer = csv.writer(file)
  writer.writerows(newStudents)
````

In the above example if the file **students.csv** already exists in your directory, the new student records are added to the end of the file. If the file doesn't exist, it is created.

The CSV library in Python does a lot of things, and we have only scratched the surface of how you can work with comma separated files with the library. For more details visit the [CSV library page in the Python documentation](https://docs.python.org/3/library/csv.html).

### **Reading and Writing to JSON Files**

The Python standard library also provides the ability to read and write to JSON files. Let's start by creating a JSON file. In Python, the structure of JSON file is typically represented as a dictionary or a list of dictionaries, as seen below:


````python
import json

students = [
  {"name": "Alex","student_id": 12345,"grade": 85},
  {"name": "Jordan","student_id": 67890,"grade": 92}
]

with open("students.json","w") as file:
  json.dump(students,file,indent=4)
````

In the above example we use the function **json.dump()** to write the contents of our list of dictionary items to the students.json file. With JSON files, maintaining readability is important, so we also specify an indentation level of 4 spaces.

If we want to read our data back from our newly created JSON file, we can use the **json.load()** function as seen below:

````python
import json

with open("students.json","r") as file:
  students = json.load(file)

  for student in students:
    print(f"{student['name']} has a student ID of {student['student_id']} and a grade of {student['grade']}.")
````

JSON files often contain subitems in their records. The example below show an example of how to create a more complex JSON file:

````python
import json

students = [
    {
        "id": 1001,
        "name": "Alex Kim",
        "grades": {
            "MATH101": 88,
            "ENG102": 92,
            "CS103": 85
        }
    },
    {
        "id": 1002,
        "name": "Jordan Patel",
        "grades": {
            "MATH101": 76,
            "ENG102": 81,
            "CS103": 90
        }
    },
    {
        "id": 1003,
        "name": "Samira Chen",
        "grades": {
            "MATH101": 93,
            "ENG102": 89,
            "CS103": 94
        }
    }
]

with open("studentGrades.json","w") as file:
  json.dump(students,file,indent=4)
````

The Python code below indicates how you would read in this slightly more complex JSON file:

````python
import json

with open("studentGrades.json","r") as file:
  students = json.load(file)

  for student in students:
    print(f"Student ID: {student['id']}")
    print(f"Name: {student['name']}")

    print("Grades:")
    print(f"MATH101: {student['grades']['MATH101']}")
    print(f"ENG102: {student['grades']['ENG102']}")
    print(f"CS103: {student['grades']['CS103']}")
    print("--------------------------------------------")
````

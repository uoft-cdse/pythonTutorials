## **Loops**

In a previous tutorial, we saw that if/else/elif statements allow us to perform certain actions if specific conditions are met.

Another common form of flow control in Python is loops, which allow us to repeat a section of code multiple times. Python has two forms of loops: **for loops** and **while loops**. **For loops** perform an action for each item in a sequence, and **while loops** repeat a block of code as long as a specified condition remains True.

For loops are particularly helpful when working with lists:


````python
numbers = [1,2,3,4,5,6]
squaredNumbers = [] # <-- here we create an empty list for our squared numbers

for number in numbers: # <-- this for loop will traverse the numbers list setting each value to the variable number
    print(number) # <-- print the variable number each time through the loop
    squaredValue = number ** 2 # <-- square the number and assign it to the variable squaredValue
    squaredNumbers.append(squaredValue) # <-- add the squared value to the squaredNumbers list

print(numbers)
print(squaredNumbers)
````

In the above example, we are using a for loop to calculate the squared value of each item in our numbers list and then add that squared value into a new list. In this example the for loop iterates through the list numbers and sets the individual values to the number variable.

Note that each of the three indented lines after the for statement are part of one code block and are run each time through the loop.

In addition to using for loops with lists, they are often also used with Python's `range()` function, to run a loop a specified amount of times.

````python
for i in range(5):
  print("Hello!")
````

**range()** is helpful function broadly in Python, as it provides a few ways to generate a series of integers as seen below:

````python
for i in range(2,10): #creates a sequence of integers starting at 2 and ending at the number before the second value
  print(i)
````

````python
for i in range(0,10,2): # creates a sequence of integers start at 0 and ending at the number before the second values but jumps in step of 2
  print(i)
````

As indicated, if you would like a certain section of code to run until a specific condition is met, a **while loop** is your best option:

````python
myNumbers = []
counter = 0

while counter < 5:
    print(counter)
    myNumbers.append(counter)
    counter = counter + 1

print(myNumbers)
````

In this example, the code block in the while loop will run while the counter variable is less than 5. Each time through the loop the value of the counter is added to our myNumbers list. The counter variable is incremented by one each time through the loop.

One thing to be careful about when using a while loop is setting a condition that never becomes false, which causes the loop to never stop. This is an infinite loop, and if you need to stop an infinite loop in JupyterHub, select the <i class="fa fa-stop"></i> stop button (the little black square beside the run button) in the menu bar will interrupt the kernel and stop the loop.

In the above example you could create an infinite loop, by removing the counter incrementation line, as this would cause the counter to always be less than five.

## **Functions**

As your Python scripts grow in complexity, you may find yourself wanting to reuse code, and functions allow you to give a name to a block of code and reuse it multiple times. Without knowing it, you have already seen a few examples of built-in functions in Python, including print() and len().

Functions allow you to optionally pass input variables called parameters to the function and return an output value from them. Let's look at an example:





````python
def squared(myNumber):
    mySquare = myNumber ** 2
    return mySquare

firstSquared = squared(4)

print(firstSquared)
print(squared(10))
````

As seen in this example, all functions start with the keyword `def` and then are followed by the name of what you would like to call the function. Within the parentheses, you can optionally pass in a variable into the function and with our example this passed variable will have the name myNumber.

Using the `return` keyword, followed by a variable name or value, allows us to use values from the function outside the function itself. A function is called by referring to the name of the function, and optionally passing in the variable in parentheses. We call the function in two ways in this example. We first set the value returned from the function to the variable firstSquared. We also call the function directly in a print statement, where we output the value of 10 squared.

Passing values to a function and returning values from a function are optional as seen below:


````python
def sayHello():
  print("Hello")

sayHello()
````

More than one variable can be passed to a function:

````python
def addTwoNumbers(number1,number2):
  total = number1 + number2
  return total

print(addTwoNumbers(3,4))
````

You can also set a default value for function parameter:

````python
def powerCalc(base,exponent=2):
  result = base ** exponent
  return result

print(powerCalc(3))
print(powerCalc(3,3))
````

In the above example, if only one value is passed to the function, the default value is used.

## **Variable Scope**

Our function examples also show the basics of variable scope in Python. Variables created within a function are only available to be used within that function. If you would like to use the value of that variable outside a function, as we have seen one way to do it is to use the `return` keyword.

Here is a simple example of how improperly referencing variables defined in functions outside a function can cause an error:

````python
def addTwo(myNumber):
    newNumber=myNumber + 2
    return newNumber

print(newNumber)
````

Note that even though we are returning the value of the newNumber variable in the function, that doesn't make the variable name usable outside the function code block. Variables assigned inside a function are considered **local** variables inside the function. Variables assigned outside a function are considered **global** variables as seen in the example below:

````python
number3 = 51

def multipleTwo(number1):
  product = number1 * number3
  return product

print(multipleTwo(10))
````

In the above example the variable `number3`, is available to be used inside the `multipleTwo` function, as it is a global variable. That said, it is a better practice to pass global variables as parameters into functions.

One complication that can arise from the above example, is using the same variable name both inside and outside a function:

````python
number3 = 100

def multipleTwo(number1):
  number3 = 50
  product = number1 * number3
  return product

print(multipleTwo(10))
````

You can see from this example, with the function the local version of the `number3` variable overrides the global version of the variable `number3` assigned outside the function.

Given the potential confusion a script like the example above can cause, it is recommended that you not reuse variable names inside a function that already exist outside the function.

## **Libraries**

So far, our examples have focused on the fundamental building blocks of Python, such as variables, functions, loops, and conditional statements. These features are built into the language and allow you to write useful programs without relying on any additional code. However, to go beyond these basics and tackle more complex or specialized tasks, Python provides a powerful collection of libraries.

Libraries are essentially collections of pre-written code that can be used to perform common tasks. Python has two types of libraries: the standard library, which comes included with your Python installation and external libraries that are installed from third parties.

In the case of both the standard library and external libraries, to use them you must first import them into your script as seen below with the math library from the standard library:



````python
import math
print(math.sqrt(16))
````

In the above example, we can only use the `sqrt()` function of the math library, after we have imported the library, using the `import` keyword. In this example, we imported the entire `math` library, but you can also import a specific piece of the library as seen below:

````python
from math import sqrt
print(sqrt(16))
````

Using libraries from the standard library requires no additional setup beyond importing them. A complete list of available standard library modules can be found on the [Python website](https://docs.python.org/3/library/index.html).

External libraries have been created by the Python community and extend the functionality of the language far beyond the standard library. External libraries often focus on specialized tasks or real-world problems and must be installed on your computer typically using a pip command as seen below:

```console
pip install Pillow
```

These pip commands can also be used in a Jupyter Notebook, but they are typically preceded by an exclamation mark:

```

```

After being installed, external libraries can be imported just like libraries from the standard library. A good list of available Python libraries can be found on the [Python Package Index](https://pypi.org/).

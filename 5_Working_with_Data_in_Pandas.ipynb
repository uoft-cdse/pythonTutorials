## **Data Structures in Pandas**

Pandas provides two useful data object types: **Series** and **DataFrames**.

Series are one-dimensional arrays of data:

```python
import pandas as pd

mySeries = pd.Series([1,2,3,4,5])
print(mySeries)
```

In the above code, when we import the Pandas library, we created the alias `pd`. Using an alias allows us to shorten our code. This alias can technically be any string, but the unofficial standard for Pandas is "pd".

When we print out a Pandas Series, the left column represents the index value of each item in the Series, and the right column represents the value of the item.

By default, when creating a Series, the items are indexed numerically, but you can also provide your own indexes:

```python
import pandas as pd

mySeries = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
print(mySeries)
```

You can also create a Series using a Python dictionary. In the below example the keys become the index, and the values become the data:

```python
import pandas as pd

data = {'apples': 5, 'oranges': 3, 'bananas': 7}

mySeries = pd.Series(data)
print(mySeries)
```

You can access items in a Series similar to how you access items in a list or dictionary:

```python
import pandas as pd

gradesSeries = pd.Series([55,78,87])
print(gradesSeries[1])

tempSeries = pd.Series([98.6, 98.9, 100.2], index = ['sample1','sample2','sample3'])
print(tempSeries['sample2'])
```

While Series can serve a useful purpose in Pandas, DataFrames which organize data in rows and columns, like a spreadsheet, are more commonly used.

DataFrames can be created in a variety of ways including from a dictionary:


```python
import pandas as pd


data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['Toronto', 'Vancouver', 'Montreal']
}

df = pd.DataFrame(data)
print(df)
```

Columns of DataFrames can be accessed using the format below:

```python
# accessing one column of data from a DataFrame
print(df['Name'])

# accessing multiple columns of data from a DataFrame
print(df[['Name', 'Age']])
```

DataFrame rows can be accessed by index name using the `.loc` property or by index position when using `.iloc`:

```python
import pandas as pd

students = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Grade': [82,99,72]
}

df = pd.DataFrame(students, index=['student1', 'student2', 'student3'])

print(df)
print("---------------------------------")
print(df.loc["student2"])
print("---------------------------------")
print(df.iloc[2])
```

## **Working with CSV Files**

Thus far we have created DataFrames within Python code itself, but you can also build DataFrames from external files like CSV files:

```python
import pandas as pd

df = pd.read_csv('UNData.csv')
print(df)
```

In the above example our CSV file's first row contained the header values of our columns.

If your CSV file does not contain the labels of your columns, you can specify them in the `.read_csv()` function call:

```python
import pandas as pd

# Reading a file with no header and assigning names manually
df = pd.read_csv('noHeaders.csv', header=None, names=['City', 'Province', 'Population'])
df.head()
```

In addition to reading from CSV files, you can also write DataFrames to CSV files using the `.to_csv()` function:

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['Toronto', 'Vancouver', 'Montreal']
}

df = pd.DataFrame(data)

df.to_csv('employees.csv', index=False)
```

In the above example, we have set the `.to_csv()` index parameter to `False`, as we do not want to include the numeric index values as the first column of our CSV file.

Pandas also allows you to filter your DataFrame based on a column value as seen below:

```python
df[df['Population 2020'] > 200000000]
```

We can also add new column:

```python
df['10 Year Change'] = df['Population 2020'] - df['Population 2010']
df.head()
```

In the above example, we added a new column called `10 Year Change`, which is created by *calculating the difference between the 2020 and 2010* population columns.

This calculation is done for every row in our DataFrame, and the result is placed in our new column.

You can also easily delete a column:

```python
df.drop(columns=['Country Code'], inplace=True)
df.head()
```

In the above example, we set the `inplace` value to True, so we can make direct changes on our DataFrame.

If you want to keep your original DataFrame, you can leave out this value or set it to False, and then create a new DataFrame as seen below:

```python
newDf = df.drop(columns=['10 Year Change'])
print("New DataFrame:")
print(newDf.head())
print("Original DataFrame")
print(df.head())
```

Pandas can also provide simple statistics on your dataset as seen below:

```python
print(df['Population 2020'].mean())
print(df['Population 2020'].min())
print(df['Population 2020'].max())
```

Pandas also allows you to sort your DataFrame by column:

```python
df.sort_values(by='Population 2020',inplace=True)
df.head()
```

Sorting is done in ascending order by default, but that can be switched to descending order:

```python
df.sort_values(by='Population 2020',ascending=False, inplace=True)
df.head()
```

## **Handling Missing Data**

Often when working with data, our datasets will contain missing values. Pandas has several built-in tools that can help us clean and manipulate this data.

To see this in action, let's load a new dataset from a CSV, which contains grocery store inventory data:

```python
import pandas as pd

df = pd.read_csv('groceries.csv')
df.head()
```

Taking a closer look at this DataFrame, you will see several rows with missing data. When a CSV file with missing data is read into Pandas by default those cells are given the value of `NaN`, which stands for **“not a number”**.

A simple way to handle missing data, is to remove any row that contains at least one missing value:

```python
cleanDf = df.dropna()
cleanDf.head()
```

The `.dropna()` function can also be used to remove rows that are only missing values in a specific column:

```python
cleanDf = df.dropna(subset=['item_name'])
cleanDf.head()
```

Pandas also allows you to replace missing values with a specific value using the `.fillna()` function :

```python
df["item_name"] = df["item_name"].fillna("Unknown Item Name")
df.head()
```

This replacement value can also be calculated from a value in another column from the same row.

In the code below, we set any missing sale item price cells to a value of 10% higher than the wholesale price:

```python
df["item_sale_price"] = df["item_sale_price"].fillna(df["item_wholesale_price"]*1.10)
df.head()
```

Pandas allows you to group data together to get summary statistics on a larger dataset. This follows a **split-apply-combine** logic:

1.   **Split** the data into groups based on some criteria
2.   **Apply** a function to each of these groups
3.   **Combine** the results into a new data structure

Pandas has several built-in functions that you can apply to grouped data including `.mean()`, `.sum()` and `.count()`.

Below is an example of how use Pandas to find the total sales for each branch contained in a DataFrame:

```python
import pandas as pd

data = {
    'Branch': ['East', 'West', 'East', 'North', 'West', 'East'],
    'Sales': [100, 200, 150, 300, 250, 50],
    'Region': ['A', 'B', 'A', 'A', 'B', 'A']
}

df = pd.DataFrame(data)

# Group by Branch and sum the Sales
branch_totals = df.groupby('Branch')['Sales'].sum()

print(branch_totals)
```

You can also apply multiple functions at one time:

```python
stats = df.groupby('Branch')['Sales'].agg(['mean', 'max', 'count'])
stats
```

Depending on the amount of columns within your results, your split-apply-combine action will either return as a Pandas Series (as in our first example) or a DataFrame (as in our second example).

You can interact with the results in the same way as you would any Series or Dataframe.

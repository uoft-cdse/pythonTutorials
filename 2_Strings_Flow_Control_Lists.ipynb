## **Comparisons and Conditionals**

The real power of programming isn't just running a series of statements from start to finish; things become more dynamic when you begin to perform certain actions based on specific conditions. This computational decision-making process is called **flow control** and allows you to define a set of rules that instruct Python to run your statements in a particular order.

We encounter flow control every day in our regular lives. For example, take the decision process behind whether you might go for walk. You may look outside to check if it is raining, and if the answer is yes, rummage through your closet to make sure you have an umbrella. If you can't find your umbrella, you decide to wait until later to go for your walk, otherwise if you did find it, you go outside and start your journey.  

Python provides a few methods for flow control, but central to all of them are comparisons, expressions that allow you to evaluate the relationship between two values.

Comparisons rely on another Python data type **Booleans**, which are used to express logical conditions and can only be either `True` or `False`:

```python
isItSnowing = True
didTheyPass = False
```

When defining a Boolean variable True or False must be capitalized, and unlike a string, the value is not contained within quotation marks.

Doing comparisons within Python also requires various operators, as listed below:

| Operator |         Meaning          |
|:--------:|:------------------------:|
|    ==    |         Equal to         |
|    !=    |       Not equal to       |
|    <     |        Less than         |
|    >     |       Greater than       |
|    <=    |  Less than or equal to   |
|    >=    | Greater than or equal to |

Now that we have a better understanding of both Booleans and comparison operators, let's look at one of Python's simplest flow control tools, **if** statements.

If statements, in their most basic form test if a certain condition is True. If that condition is met, a series of instructions will be performed:

```python
age = 50
if age > 40:
  print("You are older than 40.")
```

Before we continue with other flow control approaches in Python, it is important to understand how code blocks are organized in the language.

Code blocks are sections of instructions that are grouped together. If you have ever done programming in other languages these blocks are often defined by brackets, but in Python this grouping is done using tabs. A line that ends with a colon, starts a new code block and all the lines that are indented to the same level are part of the same code block. When the indentation goes back that code block ends. Let's look at an example:

```python
temperature = 27

if temperature < 0: # the if statement code blocks starts here
  print("The temperature is below zero.") # this is part of the code block in the if statement
  print("It is probably cold outside.") # this is part of the code block in the if statement

print("This is a third print statement.") # this is NOT part of the code block for the if statement
```

In the above example, you can see that the two lines after the if statement are indented to the same level and are considered to be part of one code block. If this statement had been found to be True, these two lines would have been printed. In this example, the condition was found to be False, as the temperature variable has a value of 27, which is greater than zero. The third print statement was run however, because since it was indented, it is not considered to be part of the code block within the if statement.

If statements work great if you only want an action to occur if something is true, but `if/else` statements are helpful if you also want something to occur when a condition tests False, as seen below:

```python
grade = 75

if grade >= 50:
  print("You passed the course.")
else:
  print("You failed the course.")
```

These `if/else` statements work well if you only want to test if one thing is true, but if you want to test multiple conditions in order, you can use `if/elif` statements:

```python
score = 75

if score >= 90:
  print("Grade: A")
elif score >= 80:
  print("Grade: B")
elif score >= 70:
  print("Grade: C")
else:
  print("Grade: D or lower")
```

Thus far we have only done simple comparisons, but you can combine comparisons using the `and` and `or` keywords. If you use the `and` keyword all of your comparisons must be true for your instructions to be performed, but in the case of `or` keyword only one comparison must test true for your instruction to be completed:

```python
age = 50
name = "Brooke"

if age > 40 and name == "Harish":
  print("You are over 40 and named Harish")

if age < 40 or name == "Brooke":
  print("You less than 40 or could be named Brooke.")
```

In this example, the first print statement isn't performed because while the age variable value is over 40, the variable name's value is not Harish, and both conditions must test true for the action to be executed because we used the `and` keyword.

The second print statement is performed because while the age variable is not less than 40, the name variable does equal Brooke. Since we used the `or` keyword in this set of comparisons, only one of the conditions needed to be true for the action to be performed.

## **Lists**

In our examples thus far, we have been using variables that consist of a single value, but Python also provides a few data types that allow you to collect multiple items together in one variable.  One of the most common is a list, which allows you to collect a sequence of related items together. Let’s look at examples of how you can create a list:

```python
scores = [10,23,100]
temperatures = [-10.1,35.4,11.23]
```

As shown above, you can make a list by placing values inside square brackets and separating them with commas. Lists in Python can contain many Python data types as items, including multiple data types in one list. A list can even contain another list. The following are all valid examples of lists:

```python
list_example1 = [1,2.3,"Toronto","Ontario"]
list_example2 = [2,[3,4,5],6,"Bob"]
list_example3 = ["English", "French", "Spanish"]
```

Lists in Python are ordered, meaning that each item has a specific position called an index. The first index of a list is 0, and you can use these indexes to access individual items in a list:

```python
myGrades = [89,72,95,64]
print(myGrades[0])
print(myGrades[3])
```

If you want to access a portion of a list, you can use a **slice**. A slice consists of two numbers separated by a colon, with the slice starting at the index of the first number and going up to, but not including the second number as seen below:

```python
fruits = ["apple","orange", "kiwi", "cherry","pear"]
print(fruits[0:3])
print(fruits[2:4])
```

You can explore a more [in-depth explanation of slicing](https://www.pythonmorsels.com/slicing/) for further background on this powerful Python tool.

Python provides a few different functions to add items to a list: `append()` will add items to the end of a list, while `insert()` will insert an item at a specific location:

```python
temperatures = [10,33,55]
temperatures.append(15)
print(temperatures)
```

```python
temperatures.insert(1,15) # insert at index 1
print(temperatures)
```

With the `insert()` function, the first value contained in the brackets is the index position that you wish to insert the value to.

Similarly, Python provides a few functions to remove items from a list, including `del`, `pop()` and `remove()`.

The `remove()` function removes the first instance of an item found in a list

```python
my_list = [1,2,3,4,3,5]
my_list.remove(3)
print(my_list)
```

The `del` keyword removes an item from a list at a specific index

```python
del my_list[3]
print(my_list)
```

Finally, the `pop()` function removes an item from a specific index from the list and allows it to be assigned to a variable.

```python
former_last_item = my_list.pop(3)
print(former_last_item)
```

```python
print(my_list)
```

It can be helpful to know the length of a list. The function `len()` helps us do that:

```python
the_list = [1,34,22,11,10]
list_length = len(the_list)
print(list_length)
```

## **Dictionaries**

As we have seen from the above examples, lists can be a very useful way to collect information together in Python. But one drawback of lists is that you must know the item's numeric position to retrieve it directly.

Python provides another data type that lets you access items via a meaningful name, **dictionaries**. This container type organizes data in key-value pairs, where the key represents a label, and the value is the information. A real-world example of a key-value pair could be seen in your contacts list which features names (a key) and phone numbers (a value).

Dictionaries are defined in Python using `{}` brackets, with the key and value separated by a colon, and the individual pairs separated by commas, as seen below:

```python
myCar = {'brand':'toyota', 'year':2006, 'model':'RAV 4'}
```

You can access an item in the dictionary by using its key:

```python
print(myCar["brand"])
print(myCar["model"])
```

Adding an item to an existing dictionary is simple:

```python
myCar["colour"] = "blue"
print(myCar)
```

If you would like to update an existing item, you can use a similar format:

```python
myCar["colour"] = "red"
print(myCar)
```

Like lists, dictionaries also support removing items, but instead of removing them by index position, you remove by key using `pop()` or `del`:

```python
del myCar["year"]
print(myCar)
```

```python
myColour = myCar.pop("colour")
print(myCar)
print(f"The colour of the car is {myColour}")
```

You can see from the above example that `pop()` both removes the value from the dictionary, and returns the value which is then set to the variable myColour.

As we saw earlier, lists can contain many different data types as items, and this includes dictionaries. A list of dictionaries is a very common way to organize data in Python:

```python
groceryOrderList = [
  {"item":"cereal", "price":1.99,"quantity":2},
  {"item":"toothpaste", "price":3.99,"quantity":1},
  {"item":"milk", "price":6.99,"quantity":2}
]

print(groceryOrderList[1]["item"])
print(groceryOrderList[2])
```

You can see from this example that to access an item in a list of dictionaries, you use two sets of square brackets. The first indicates the index of the list item, and the second specifies the key of the dictionary item you would like to access.

This example also shows a small quirk with Python. Usually with Python, one line of code equals one line of text, but there is an exception when values are contained within brackets. This rule allows us to write our list definition in a bit more of a readable format than if all our items were in one long line.

Dictionaries also introduce another potential Python error type a **KeyError**:

```python
courseDetails = {"courseName":"Introduction to Chemistry", "courseCode":"CHEM101", "courseCapacity":120}
print(courseDetails["enrollment"])
```

In the above example, we tried to print an item from our `courseDetails` dictionary with a key that did not exist in the dictionary causing an error.

## **String Manipulation**

Python provides several useful ways to manipulate strings, including the ability to change the case of a string as seen below:

```python
sentence = "this is a lowercase sentence"
print(sentence.upper())

sentence = "THIS IS AN UPPERCASE SENTENCE"
print(sentence.lower())

sentence = "this is a sentence with a title"
print(sentence.title())
```

Python also makes it easy to split up a string into a list of strings using the `.split()` method:

```python
sentence = "this is a sentence"
print(sentence.split())
```

The default behaviour is to split by whitespace, but you can also specify a delimiter by including a string within the split function call:

```python
sentence = "this,is,a,sentence"
print(sentence.split(","))
```

Additionally, the `.join()` method allows you to concatenate a list of strings into a new string:

```python
myStringList = ["this","is","a","sentence"]
print(" ".join(myStringList))
```

In the above example, a space is placed between each string in our new string, but you can specify any string to be placed between the strings from your list:

```python
myStringList = ["this", "is", "another","sentence"]
print("-".join(myStringList))
print("space".join(myStringList))
```

Often when working with string data, you may have unwanted text, such as whitespace, surrounding your string. The `.strip()` method can help remove these characters:

```python
myMessyString = "             My sentence has spaces before and after it.        "
myCleanString = myMessyString.strip()
print(myCleanString)
```

Again the default behavior will remove whitespace from the beginning and end of your string, but you can specify a string to remove when using `.strip()`:

```python
myMessyString = "-----My sentence has dashes before and after it-----"
myCleanString = myMessyString.strip("-")
print(myCleanString)
```

Python provides many more ways to manipulate strings, consult the [Python String documentation](https://docs.python.org/3/library/string.html) for more details.  

## **Escaping Characters**

In Python certain characters have special meanings, and need to be **escaped** when using them in a string. You can escape these special characters by using a backslash `\\` before the character. Potential special characters include single quotes `'` and double quotes `"`, the backslash itself `\\` and special formatting characters such as newline `\n` and tab `\t`.

Below are few examples of how to escape characters:

```python
print("This string has a backslash (\\) in it ")
print("This is on the first line.\nThis is on a second line.")
print("Here is some text.\tThis text continues after a tab.")
print("Here is a string contained within double quotes. I want to use a \" in it")
print('Here is a string contained within single quotes. I want to use a \' in it')
```

To use the same quotation mark as what the string is contained by within in the string itself, we must escape the quotation mark.
